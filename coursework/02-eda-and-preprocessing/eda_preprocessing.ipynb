```python
import numpy as np
```

```python
edades
```

```python
import matplotlib.pyplot as plt

fare_array = np.array(df['Fare'])
```

```python
fig, ax = plt.subplots()
ax.set_title('Boxplot')
ax.boxplot(edades, labels = ['edades'])
```

```python
q1 = np.quantile(edades, 0.25)
q2 = np.quantile(edades, 0.5)
q3 = np.quantile(edades, 0.75)
```

```python
q1, q2, q3
```

```python
iqr = q3 - q1
```

```python
sup = q3 + 1.5 * iqr
inf = q1 - 1.5 * iqr
# criterio de Tukey
```

```python
sup, inf
```

```python
outliers_indices = np.where((edades > sup) | (edades < inf))
```

```python
outliers_indices
```

```python
edades = np.delete(edades, outliers_indices)
```

```python
fig, ax = plt.subplots()
ax.set_title('Boxplot')
ax.boxplot(edades, labels = ['edades'])
```

```python
df[['age', 'bmi', 'children', 'charges']]
```

```python
df[numeric_columns]
```

```python
for col in numeric_columns:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  sup = q3 + 1.5 * iqr
  inf = q1 - 1.5 * iqr
  outliers_indices = df[(df[col] > sup) | (df[col] < inf)].index
  df = df.drop(outliers_indices)
```
